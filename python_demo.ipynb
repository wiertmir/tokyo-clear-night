```python
@dataclass
class City:
    name: str
    population: int = 0
    tags: list[str] = field(default_factory=list)

    @property
    def label(self) -> str:
        return f"{self.name} ({self.population:,})"

    @classmethod
    def tokyo(cls) -> "City":
        return cls("Tokyo", 14_000_000)

    def rename(self, new_name: str) -> None:
        self.name = new_name

    def __repr__(self) -> str:
        return self.label


class Megacity(City):
    def __init__(self, name: str) -> None:
        super().__init__(name, 10_000_000)
        self.visits = 0


osaka = Megacity("Osaka")
osaka.rename("Ōsaka")
print(City.tokyo(), osaka, osaka.visits)
```

## Control flow, exceptions & async

| Element | Color | Source |
|---|---|---|
| <span style="color:#bb9af7;font-family:monospace">if  for  match  case  try  except  finally  raise</span> | <code>#bb9af7</code> | keyword.control |
| <span style="color:#bf4f51;font-family:monospace">ValueError  KeyError</span> | <code>#bf4f51</code> | support.type.exception (Blushed Brick); Pylance overrides with built-in #0db9d7 |
| <span style="color:#6f00ff;font-family:monospace">global</span> | <code>#6f00ff</code> | storage.modifier.declaration (Electric Indigo) |
| <span style="color:#bb9af7;font-family:monospace">async  await</span> | <code>#bb9af7</code> | storage.type / keyword |
| <span style="color:#89ddff;font-family:monospace">:=</span> | <code>#89ddff</code> | keyword.operator |

```python
counter = 0


def bump() -> None:
    global counter
    counter += 1


def describe(value: object) -> str:
    match value:
        case int(n) if n > 0:
            return "positive int"
        case [first_item, *rest]:
            return f"list starting with {first_item}"
        case _:
            return "something else"


try:
    if (n := len(os.sep)) > 0:
        raise ValueError(f"separator length {n}")
except (ValueError, KeyError) as err:
    print("caught:", err)
finally:
    for _ in range(2):
        bump()


async def fetch() -> str:
    return "done"

print(describe(5), describe([1, 2]), await fetch(), counter)
```
